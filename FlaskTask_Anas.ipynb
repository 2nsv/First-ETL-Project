### Transformations

```python
customers
```

```python
customers = customers.drop_duplicates()
customers["age"] = pd.to_numeric(customers["age"])
customers["email"] = customers["email"].fillna("no email provided")
customers["full_name"] = customers["first_name"] + " " + customers["last_name"]
```

```python
orders
```

```python
orders = orders.drop_duplicates()
orders["quantity"]=pd.to_numeric(orders["quantity"])
orders["order_date"] = pd.to_datetime(orders["order_date"])
```

```python
Products
```

```python
Products = Products.drop_duplicates()
Products["unit_price"] = pd.to_numeric(Products["unit_price"])
```

```python
Sales = pd.merge(customers, orders, on="customer_id" , how="inner")
Sales = pd.merge(Sales , Products , on="product_id" , how = "inner")

Sales
```

```python
Sales["total_amount"] = Sales["quantity"] * Sales["unit_price"]
```

```python
Final_df = Sales[["order_id" ,"full_name","product_name" , "quantity" ,"unit_price" , "total_amount" ]]
```

### Load (SQL Server)

```python
data = list(Final_df.itertuples(index=False, name=None))
```

```python
import pyodbc
```

```python
conn = pyodbc.connect(
    "Driver={ODBC Driver 17 for SQL Server};"
    "server=.\SQLEXPRESS;"
    "Database=FlaskDB;"
    "trusted_connection=yes;"
)
cursor = conn.cursor()
cursor.execute(
'''
CREATE TABLE Sales (

   order_id           INT ,
   full_name          VARCHAR(50)  ,
   product_name       VARCHAR(50)  ,
   quantity           INT ,
   unit_price         INT ,
   total_amount       INT 
)

'''
)        

conn.commit()
cursor.close()
conn.close()
```

### Flask Dashboard

```python
from flask import Flask , render_template

app = Flask(__name__)

@app.route("/")
def dashboard() :
 
 conn = pyodbc.connect(
    "Driver={ODBC Driver 17 for SQL Server};"
    "server=.\SQLEXPRESS;"
    "Database=FlaskDB;"
    "trusted_connection=yes;"
 )
 cursor = conn.cursor()
 cursor.execute(
 '''
 SELECT * FROM Sales; 
 
 '''
 )   
 columns = [column[0] for column in cursor.description]
 rows = cursor.fetchall()     
 cursor.close()
 conn.close()

 return render_template(
    "dashboard.html",
    columns = columns , 
    rows = rows
)


if __name__ == "__main__":
    app.run(
        debug=True,
        use_reloader=False
    )
```
